```python
def calculate_features(
    air_temperature,
    process_temperature,
    rotational_speed,
    torque,
    tool_wear
):
    """
    Create engineered features from machine sensor values.
    """

    temperature_difference = (
        process_temperature - air_temperature
    )

    power = (
        torque * rotational_speed
    )

    torque_toolwear = (
        torque * tool_wear
    )

    return {
        "Temperature_Difference": temperature_difference,
        "Power": power,
        "Torque_ToolWear": torque_toolwear
    }


def encode_machine_type(machine_type):
    """
    Encode machine type into Type_H and Type_M.
    """

    type_h = 1 if machine_type == "H" else 0
    type_m = 1 if machine_type == "M" else 0

    return type_h, type_m


def get_risk_level(probability):
    """
    Convert failure probability into risk level
    and maintenance recommendation.
    """

    if probability >= 0.70:
        return (
            "HIGH",
            "Maintenance required immediately."
        )

    elif probability >= 0.40:
        return (
            "MEDIUM",
            "Schedule maintenance soon."
        )

    else:
        return (
            "LOW",
            "Machine condition looks normal."
        )
```
